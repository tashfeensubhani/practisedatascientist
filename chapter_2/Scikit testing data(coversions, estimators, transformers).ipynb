<h1>Manually splitting test data</h1>

```python
def split_train_test(data, test_ratio):
    shuffled_indices = np.random.permutation(len(dt))
    test_size = int(len(dt) * test_ratio)
    test_indices = shuffled_indices[:test_size]
    train_indices = shuffled_indices[test_size:]
    np.random.seed(42)
    return dt.iloc[train_indices], dt.iloc[test_indices]
```

```python
train_set, test_set = split_train_test(dt, 0.2)
```
