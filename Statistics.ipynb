# 📊 Pearson’s Correlation (r) – Machine Learning Notes

## 🔍 What is Correlation (Pearson’s r)?
**Correlation** measures the **strength and direction** of a **linear relationship** between two continuous variables.

> **Pearson’s r** is the most commonly used correlation coefficient.

---

## 📐 Formula of Pearson’s r

 Google!! or notes
 

- \( \bar{x}, \bar{y} \): Mean of x and y
- \( r \) ranges from **-1** to **+1**

---

## 🔁 Interpretation Table

| **r Value**     | **Strength**     | **Direction**     |
|------------------|------------------|--------------------|
| **1.0**           | Perfect           | Positive (↑↑)       |
| **0.7 – 0.9**     | Strong            | Positive (↑↑)       |
| **0.4 – 0.6**     | Moderate          | Positive (↑↑)       |

| **0**            | None              | No Relationship     |
| **-0.1 – -0.3**   | Weak              | Negative (↑↓)       |
| **-0.4 – -0.6**   | Moderate          | Negative (↑↓)       |
| **-0.7 – -0.9**   | Strong            | Negative (↑↓)       |
| **-1.0**          | Perfect           | Negative (↑↓)       |

## ✅ Uses of Pearson’s Correlation

| **Use Case**                    | **Purpose**                                           |
|----------------------------------|--------------------------------------------------------|
| Feature Selection (ML)          | Identify redundant or related features                 |
| Exploratory Data Analysis (EDA) | Understand relationships in data                       |
| Finance                         | Analyze stock price movement correlations              |
| Medical Research                | Measure effect of drug dose vs. response               |
| Sensor Systems                  | Detect relationships between sensor outputs            |

---

## ⚠️ When to Use Pearson’s r

Use Pearson’s correlation when:
- Both variables are **numeric & continuous**
- Relationship is **linear**
- No strong **outliers**
- Data is **normally distributed**

---

## ❌ When NOT to Use Pearson’s r

Avoid it when:
- Data is **categorical**
- Relationship is **non-linear**
- **Outliers** are present
- Use **Spearman** or **Kendall** for non-linear data

---

## 🔁 Alternatives to Pearson’s r

| **Method**        | **Best For**                      |
|------------------|-----------------------------------|
| Spearman Rank     | Non-linear, ordinal data          |
| Kendall Tau       | Small datasets or many tied ranks |
