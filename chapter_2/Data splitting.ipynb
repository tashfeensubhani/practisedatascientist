 # Data splitting:
         is defined as divide data into 3 parts: test, train and validation...
  ## What is Training Set, Validation Set, Testing Set:

    1 **Training Set:**
             Use to train ML model, this is the core data set where the model learns to understand patterns and relationship in data.
    2 **Validation Set:**
             Assists in tuning the model. It evakuates the model's performance during the training phase, gelping to adjust hyper-parameter and prevent                 overfitting.
    3 **Test Set:**
             Provides a fair evolution of the model's performance on unseen data
